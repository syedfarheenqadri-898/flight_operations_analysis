## 9. Grouping and Aggregation

Grouping allows us to compare flight data across categories such as airlines. Aggregation helps calculate values such as average delay, total passengers, and average ticket price.

```python
df.groupby("Airline")["Delay_Minutes"].mean().sort_values(ascending=False)
```

```python
df.groupby("Airline")["Average_Ticket_Price"].mean().sort_values(ascending=False)
```

```python
df.groupby("Airline")["Passengers"].sum().sort_values(ascending=False)
```

```python
df.groupby("Airline").agg({
    "Passengers": "sum",
    "Delay_Minutes": "mean",
    "Average_Ticket_Price": "mean"
}).sort_values("Passengers", ascending=False)
```

## 10. Data Transformation using apply() and Lambda

The `apply()` function allows us to apply an operation to a column. A lambda function is used to create a simple transformation without defining a separate function.

```python
df["Delay_Category"] = df["Delay_Minutes"].apply(
    lambda x: "Delayed" if x > 15 else "On Time"
)
```

```python
df[["Delay_Minutes", "Delay_Category"]].head(10)
```

```python
df["Delay_Category"].value_counts()
```

## 11. Combining DataFrames

Pandas provides `merge()`, `join()`, and `concat()` to combine data from multiple DataFrames.

## 1.Merge()

```python
airline_summary = df.groupby("Airline", as_index=False)["Passengers"].sum()

airline_summary
```

```python
merged_df = df.merge(airline_summary, on="Airline", suffixes=("", "_Total"))

merged_df.head()
```

## 2.Join()

```python
airline_delay = df.groupby("Airline")["Delay_Minutes"].mean().to_frame("Average_Delay")
```

```python
df.join(airline_delay, on="Airline").head()
```

## 3. Concat()

```python
part1 = df.iloc[:5]
part2 = df.iloc[5:10]

combined_df = pd.concat([part1, part2])

combined_df
```

## 12. Key Analysis Questions

The following questions are used to identify important patterns and trends in the Flight Operations dataset.

```python
# 1. Average delay by airline
print("Average Delay by Airline:")
display(df.groupby("Airline")["Delay_Minutes"].mean().sort_values(ascending=False))
# 2. Total passengers by airline
print("Total Passengers by Airline:")
display(df.groupby("Airline")["Passengers"].sum().sort_values(ascending=False))
# 3. Number of flights by month
print("Number of Flights by Month:")
display(df["Month"].value_counts().sort_index())
# 4. Average delay by weather condition
print("Average Delay by Weather:")
display(df.groupby("Weather")["Delay_Minutes"].mean().sort_values(ascending=False))
# 5. Average ticket price by travel class
print("Average Ticket Price by Travel Class:")
display(df.groupby("Travel_Class")["Average_Ticket_Price"].mean().sort_values(ascending=False))
# 6. Flight status distribution
print("Flight Status:")
display(df["Flight_Status"].value_counts())
```

## 13. Key Findings and Observations

The analysis of the Flight Operations dataset reveals several patterns related to airline performance, passenger volume, delays, pricing, weather conditions, and flight status.

```python
highest_delay_airline = df.groupby("Airline")["Delay_Minutes"].mean().idxmax()
highest_delay = df.groupby("Airline")["Delay_Minutes"].mean().max()

highest_passenger_airline = df.groupby("Airline")["Passengers"].sum().idxmax()
highest_passengers = df.groupby("Airline")["Passengers"].sum().max()

most_common_weather = df["Weather"].value_counts().idxmax()
most_common_status = df["Flight_Status"].value_counts().idxmax()

highest_delay_weather = df.groupby("Weather")["Delay_Minutes"].mean().idxmax()
highest_delay_weather_value = df.groupby("Weather")["Delay_Minutes"].mean().max()

highest_price_class = df.groupby("Travel_Class")["Average_Ticket_Price"].mean().idxmax()
highest_price = df.groupby("Travel_Class")["Average_Ticket_Price"].mean().max()

print("Airline with highest average delay:", highest_delay_airline, round(highest_delay, 2), "minutes")
print("Airline with most passengers:", highest_passenger_airline, highest_passengers)
print("Most common weather condition:", most_common_weather)
print("Most common flight status:", most_common_status)
print("Weather with highest average delay:", highest_delay_weather, round(highest_delay_weather_value, 2), "minutes")
print("Travel class with highest average ticket price:", highest_price_class, round(highest_price, 2))
```

## 13. Key Findings and Observations

1. IndiGo had the highest average delay at approximately 25.03 minutes, followed by SpiceJet at about 24.14 minutes.

2. Vistara carried the highest total number of passengers, with 4,484 passengers across the recorded flights.

3. Rain was the most frequently recorded weather condition, appearing in 43 flights.

4. Most flights were On Time (114 flights), while 59 were Delayed and 7 were Cancelled.

5. Flights during Rain had the highest average delay at approximately 22.22 minutes, suggesting that weather conditions may influence flight delays.

6. Business Class had the highest average ticket price at approximately 16,780, followed by Premium Economy and Economy.

7. The analysis shows that grouping and aggregation can be used to compare airlines, weather conditions, passenger volumes, delays, and ticket prices.

8. The dataset demonstrates how Pandas can be used to transform raw flight records into meaningful information for understanding flight operations.
